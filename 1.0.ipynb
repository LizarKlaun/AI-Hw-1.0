```python
temperatures = [2, 5, 4, 6, 7, 3, 1]

avg_temp = sum(temperatures) / len(temperatures)
max_temp = max(temperatures)
min_temp = min(temperatures)
range_temp = max_temp - min_temp

print(f"Середнє значення: {avg_temp:.2f}")
print(f"Максимальне значення: {max_temp}")
print(f"Мінімальне значення: {min_temp}")
print(f"Різниця (розмах): {range_temp}")
```

```python
expenses = [1200, 1500, 800, 600, 1000, 400, 900]

total_expenses = sum(expenses)
avg_expenses = total_expenses / len(expenses)
max_expense = max(expenses)
min_expense = min(expenses)
days_above_1000 = sum(1 for x in expenses if x > 1000)

print(f"Загальна сума: {total_expenses}")
print(f"Середнє значення: {avg_expenses:.2f}")
print(f"Максимальна витрата: {max_expense}")
print(f"Мінімальна витрата: {min_expense}")
print(f"Днів з витратами > 1000: {days_above_1000}")
```

```python
from collections import Counter

votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]

# Підрахунок голосів
vote_counts = Counter(votes)

# Визначення лідера
leader, leader_votes = vote_counts.most_common(1)[0]

# Отримання топ-3
top_3 = vote_counts.most_common(3)

print(f"Лідер: {leader} ({leader_votes} голосів)\n")
print("Усі результати:")
for fruit, count in vote_counts.items():
    print(f"  {fruit}: {count}")

print("\nТоп-3:")
for rank, (fruit, count) in enumerate(top_3, start=1):
    print(f"  {rank}. {fruit}: {count}")
```

```python
scores = {
    "math": [80, 92, 75, 60, 88],
    "physics": [70, 65, 90, 85, 78],
    "english": [95, 100, 85, 90, 87]
}

stats = {}
for subject, marks in scores.items():
    stats[subject] = {
        "avg": sum(marks) / len(marks),
        "min": min(marks),
        "max": max(marks)
    }

# Пошук предмета з найвищим середнім результатом кодом
best_subject = max(stats, key=lambda s: stats[s]["avg"])
print(f"Предмет з найвищим середнім результатом: {best_subject}")
```
